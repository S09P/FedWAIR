```python
import os, glob, random, copy, warnings, itertools
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
warnings.filterwarnings('ignore')

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import accuracy_score, balanced_accuracy_score, confusion_matrix, roc_auc_score

print('Imports done.')

DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
if DEVICE.type == 'cuda':
    torch.backends.cudnn.benchmark = True
    print(f'Using GPU: {torch.cuda.get_device_name(0)}')
else:
    print('No GPU detected — running on CPU.')
```

```python
class FairTradeMLP(nn.Module):
    
    def __init__(self, input_dim, hidden=64):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(input_dim, hidden), nn.ReLU(),
            nn.Linear(hidden, hidden // 2), nn.ReLU(),
            nn.Linear(hidden // 2, hidden // 4), nn.ReLU(),
            nn.Linear(hidden // 4, 1), nn.Sigmoid()
        )
    def forward(self, x):
        return self.net(x)


def fairness_metrics(y_true, y_pred, protected, y_score=None):

    y_true = np.array(y_true); y_pred = np.array(y_pred); protected = np.array(protected)
    priv_idx = protected == 1
    unpriv_idx = protected == 0
    p_priv   = np.mean(y_pred[priv_idx] == 1)   if priv_idx.sum()   > 0 else 0.0
    p_unpriv = np.mean(y_pred[unpriv_idx] == 1) if unpriv_idx.sum() > 0 else 0.0
    SPD = p_unpriv - p_priv
    DI  = 1.0 - (p_unpriv / p_priv) if p_priv > 1e-9 else np.nan

    def rates(yt, yp):
        if len(yt) == 0:
            return dict(tpr=0.0, fpr=0.0, tnr=0.0, fnr=0.0, precision=0.0, f1=0.0, acc=0.0)
        tn, fp, fn, tp = confusion_matrix(yt, yp, labels=[0, 1]).ravel()
        tpr = tp / (tp + fn) if (tp + fn) > 0 else 0.0
        fpr = fp / (fp + tn) if (fp + tn) > 0 else 0.0
        tnr = tn / (tn + fp) if (tn + fp) > 0 else 0.0
        fnr = fn / (fn + tp) if (fn + tp) > 0 else 0.0
        precision = tp / (tp + fp) if (tp + fp) > 0 else 0.0
        f1 = 2 * precision * tpr / (precision + tpr) if (precision + tpr) > 0 else 0.0
        acc_g = (tp + tn) / (tp + tn + fp + fn) if (tp + tn + fp + fn) > 0 else 0.0
        return dict(tpr=tpr, fpr=fpr, tnr=tnr, fnr=fnr, precision=precision, f1=f1, acc=acc_g)

    r_priv    = rates(y_true[priv_idx],   y_pred[priv_idx])
    r_unpriv  = rates(y_true[unpriv_idx], y_pred[unpriv_idx])
    r_overall = rates(y_true, y_pred)

    EOD  = r_unpriv['tpr'] - r_priv['tpr']
    AOD  = ((r_unpriv['fpr'] - r_priv['fpr']) + (r_unpriv['tpr'] - r_priv['tpr'])) / 2
    FPRD = r_unpriv['fpr'] - r_priv['fpr']
    TNRD = r_unpriv['tnr'] - r_priv['tnr']
    PPD  = r_unpriv['precision'] - r_priv['precision']

    out = {
        'DI': DI, 'SPD': SPD, 'EOD': EOD, 'AOD': AOD, 'FPRD': FPRD, 'TNRD': TNRD, 'PPD': PPD,
        'tpr_priv': r_priv['tpr'],   'tpr_unpriv': r_unpriv['tpr'],
        'fpr_priv': r_priv['fpr'],   'fpr_unpriv': r_unpriv['fpr'],
        'tnr_priv': r_priv['tnr'],   'tnr_unpriv': r_unpriv['tnr'],
        'fnr_priv': r_priv['fnr'],   'fnr_unpriv': r_unpriv['fnr'],
        'precision_priv': r_priv['precision'], 'precision_unpriv': r_unpriv['precision'],
        'f1_priv': r_priv['f1'],     'f1_unpriv': r_unpriv['f1'],
        'f1_overall': r_overall['f1'], 'precision_overall': r_overall['precision'],
    }
    if y_score is not None:
        y_score = np.array(y_score)
        try:
            out['auc_overall'] = roc_auc_score(y_true, y_score)
        except ValueError:
            out['auc_overall'] = np.nan
        try:
            out['auc_priv'] = (roc_auc_score(y_true[priv_idx], y_score[priv_idx])
                                if priv_idx.sum() > 0 and len(np.unique(y_true[priv_idx])) > 1 else np.nan)
        except ValueError:
            out['auc_priv'] = np.nan
        try:
            out['auc_unpriv'] = (roc_auc_score(y_true[unpriv_idx], y_score[unpriv_idx])
                                  if unpriv_idx.sum() > 0 and len(np.unique(y_true[unpriv_idx])) > 1 else np.nan)
        except ValueError:
            out['auc_unpriv'] = np.nan
    return out


def full_eval(model, X_tensor, y_np, prot_np, return_scores=False):
    model.eval()
    with torch.no_grad():
        probs = model(X_tensor).squeeze().cpu().numpy()
    preds = (probs >= 0.5).astype(float)
    acc  = accuracy_score(y_np, preds)
    bacc = balanced_accuracy_score(y_np, preds)
    fm   = fairness_metrics(y_np.astype(int), preds.astype(int), prot_np, y_score=probs)
    if return_scores:
        return acc, bacc, fm, probs
    return acc, bacc, fm

print('Model / metric definitions ready.')
```

```python


def _dp_violation_penalty(probs, s, budget_e):
    
    p_overall = probs.mean()
    mask1 = s == 1
    mask0 = s == 0
    p1 = probs[mask1].mean() if mask1.any() else p_overall
    p0 = probs[mask0].mean() if mask0.any() else p_overall
    viol1 = (p1 - p_overall).abs() - budget_e
    viol0 = (p0 - p_overall).abs() - budget_e
    relu_viol = Fnn.relu(torch.stack([viol1, viol0]))
    return relu_viol.norm(p=2)


def _client_local_train(model, X_c, y_c, s_c, lr, zeta, local_epochs, batch_size, budget_e):
    
    model.train()
    opt = optim.Adam(model.parameters(), lr=lr)
    loader = DataLoader(TensorDataset(X_c, y_c, s_c), batch_size=batch_size, shuffle=True, drop_last=False)
    for _ in range(local_epochs):
        for bX, by, bs in loader:
            if bX.size(0) < 1:
                continue
            bX, by, bs = bX.to(DEVICE), by.to(DEVICE), bs.to(DEVICE)
            probs = model(bX).squeeze(-1)
            task_loss = Fnn.binary_cross_entropy(probs, by)
            fair_pen = _dp_violation_penalty(probs, bs, budget_e)
            total_loss = task_loss + zeta * fair_pen
            opt.zero_grad()
            total_loss.backward()
            opt.step()


def _client_local_metrics(model, X_c, y_c, s_c):
    
    model.eval()
    with torch.no_grad():
        probs = model(X_c.to(DEVICE)).squeeze(-1).cpu().numpy()
    preds = (probs >= 0.5).astype(int)
    y_np = y_c.numpy().astype(int)
    s_np = s_c.numpy().astype(int)
    bacc = balanced_accuracy_score(y_np, preds)
    priv = s_np == 1; unpriv = s_np == 0
    p_priv = preds[priv].mean() if priv.sum() > 0 else 0.0
    p_unpriv = preds[unpriv].mean() if unpriv.sum() > 0 else 0.0
    disc_score = p_priv - p_unpriv   # eq. 2: DP = P(y=1|S=1) - P(y=1|S=0)  (paper's own sign convention)
    return bacc, disc_score


def _propose_bo_candidate(train_x_unit, train_y, ref_point):
    
    try:
        train_x_t = torch.tensor(train_x_unit, dtype=BOTORCH_DTYPE)
        train_y_t = torch.tensor(train_y, dtype=BOTORCH_DTYPE)
        models = [SingleTaskGP(train_x_t, train_y_t[:, i:i+1]) for i in range(train_y_t.shape[1])]
        model_list = ModelListGP(*models)
        mll = SumMarginalLogLikelihood(model_list.likelihood, model_list)
        fit_gpytorch_mll(mll)

        ref_point_t = torch.tensor(ref_point, dtype=BOTORCH_DTYPE)
        partitioning = FastNondominatedPartitioning(ref_point=ref_point_t, Y=train_y_t)
        sampler = SobolQMCNormalSampler(sample_shape=torch.Size([32]))
        acq = qExpectedHypervolumeImprovement(
            model=model_list, ref_point=ref_point_t, partitioning=partitioning, sampler=sampler
        )
        bounds = torch.tensor([[0.0, 0.0], [1.0, 1.0]], dtype=BOTORCH_DTYPE)
        candidate, _ = optimize_acqf(
            acq_function=acq, bounds=bounds, q=1, num_restarts=5, raw_samples=64
        )
        return candidate.squeeze(0).numpy()
    except Exception:
        return np.random.uniform(0.0, 1.0, size=2)


def run_fairtrade_pipeline(df, protected_attr, target_col,
                            num_clients=NUM_CLIENTS, num_rounds=NUM_ROUNDS_MAIN,
                            local_epochs=LOCAL_EPOCHS, batch_size=BATCH_SIZE,
                            budget_e=FAIRNESS_BUDGET_E, lr_bounds=LR_BOUNDS, zeta_bounds=ZETA_BOUNDS,
                            n_bo_iters=N_BO_ITERS, bo_trial_epochs=BO_TRIAL_LOCAL_EPOCHS,
                            seed=42):
    
    torch.manual_seed(seed)
    np.random.seed(seed)
    random.seed(seed)

    if DEVICE.type == 'cuda':
        torch.cuda.manual_seed_all(seed)

    train_val_df, test_df = train_test_split(
        df,
        test_size=0.15,
        random_state=seed,
        shuffle=True
    )

    val_fraction = 0.15 / 0.85

    train_df, val_df = train_test_split(
        train_val_df,
        test_size=val_fraction,
        random_state=seed,
        shuffle=True
    )

    
    train_df = train_df.sample(
        frac=1,
        random_state=seed
    ).reset_index(drop=True)

    print(
        f"FairTrade split: "
        f"train={len(train_df)} ({len(train_df)/len(df):.2%}), "
        f"val={len(val_df)} ({len(val_df)/len(df):.2%}), "
        f"test={len(test_df)} ({len(test_df)/len(df):.2%})"
    )

    
    client_idx_splits = np.array_split(
        np.arange(len(train_df)),
        num_clients
    )

    clients_raw = [
        train_df.iloc[idx].reset_index(drop=True)
        for idx in client_idx_splits
    ]

    client_n = [len(c) for c in clients_raw]
    n_total = sum(client_n)

    

    X_val_np = val_df.drop(
        columns=[target_col]
    ).values.astype(np.float32)

    y_val_np = val_df[target_col].values.astype(np.float32)

    prot_val = val_df[protected_attr].values

    X_val_tensor = torch.tensor(X_val_np).to(DEVICE)

    
    X_test_np = test_df.drop(
        columns=[target_col]
    ).values.astype(np.float32)

    y_test_np = test_df[target_col].values.astype(np.float32)

    prot_test = test_df[protected_attr].values

    X_test_tensor = torch.tensor(X_test_np).to(DEVICE)

    input_dim = X_train_input_dim = train_df.drop(
        columns=[target_col]
    ).shape[1]

   

    def to_unit(lr, zeta):
        lo, hi = lr_bounds
        x0 = (lr - lo) / (hi - lo)

        lo, hi = zeta_bounds
        x1 = (zeta - lo) / (hi - lo)

        return np.clip(
            [x0, x1],
            0.0,
            1.0
        )

    def from_unit(x):
        lo, hi = lr_bounds
        lr = lo + x[0] * (hi - lo)

        lo, hi = zeta_bounds
        zeta = lo + x[1] * (hi - lo)

        return float(lr), float(zeta)

    

    global_model = FairTradeMLP(input_dim).to(DEVICE)

    
    client_tensors = []

    for cdf in clients_raw:

        X_c = torch.tensor(
            cdf.drop(columns=[target_col])
               .values.astype(np.float32)
        )

        y_c = torch.tensor(
            cdf[target_col]
               .values.astype(np.float32)
        )

        s_c = torch.tensor(
            cdf[protected_attr]
               .values.astype(np.float32)
        )

        client_tensors.append(
            (X_c, y_c, s_c)
        )

    
    cur_lr = lr_bounds[0] * 5
    cur_zeta = zeta_bounds[1] / 2.0

    
    ref_point = np.array([0.0, -1.0])

   

    round_history = []

    g_acc = None
    g_bacc = None
    g_fm = None
    g_probs = None

    

    for rnd in range(1, num_rounds + 1):

       
        local_states = []

        for cid in range(num_clients):

            model_k = FairTradeMLP(input_dim).to(DEVICE)

            model_k.load_state_dict(
                global_model.state_dict()
            )

            X_c, y_c, s_c = client_tensors[cid]

            _client_local_train(
                model_k,
                X_c,
                y_c,
                s_c,
                cur_lr,
                cur_zeta,
                local_epochs,
                batch_size,
                budget_e
            )

            local_states.append(
                copy.deepcopy(
                    model_k.state_dict()
                )
            )

        

        weights = (
            np.array(client_n, dtype=np.float64)
            / n_total
        )

        def agg_state(states):
            out = {}

            for key in states[0]:

                out[key] = sum(
                    weights[i] * states[i][key].float()
                    for i in range(num_clients)
                )

            return out

        global_model.load_state_dict(
            agg_state(local_states)
        )

        

        bas = []
        discs = []

        for cid in range(num_clients):

            m_eval = FairTradeMLP(input_dim).to(DEVICE)

            m_eval.load_state_dict(
                global_model.state_dict()
            )

            X_c, y_c, s_c = client_tensors[cid]

            bacc_k, disc_k = _client_local_metrics(
                m_eval,
                X_c,
                y_c,
                s_c
            )

            bas.append(bacc_k)
            discs.append(disc_k)

        BA_g = float(
            np.dot(weights, bas)
        )

        disc_g = float(
            np.dot(weights, discs)
        )

       

        obs_x = [
            to_unit(cur_lr, cur_zeta)
        ]

        obs_y = [
            [BA_g, -abs(disc_g)]
        ]

        for _ in range(n_bo_iters):

            cand_unit = _propose_bo_candidate(
                np.array(obs_x),
                np.array(obs_y),
                ref_point
            )

            trial_lr, trial_zeta = from_unit(
                cand_unit
            )

            trial_bas = []
            trial_discs = []

            for cid in range(num_clients):

                model_trial = FairTradeMLP(
                    input_dim
                ).to(DEVICE)

                
                model_trial.load_state_dict(
                    global_model.state_dict()
                )

                X_c, y_c, s_c = client_tensors[cid]

                _client_local_train(
                    model_trial,
                    X_c,
                    y_c,
                    s_c,
                    trial_lr,
                    trial_zeta,
                    bo_trial_epochs,
                    batch_size,
                    budget_e
                )

                bacc_k, disc_k = _client_local_metrics(
                    model_trial,
                    X_c,
                    y_c,
                    s_c
                )

                trial_bas.append(bacc_k)
                trial_discs.append(disc_k)

            trial_BA = float(
                np.dot(weights, trial_bas)
            )

            trial_disc = float(
                np.dot(weights, trial_discs)
            )

            obs_x.append(cand_unit)

            obs_y.append(
                [
                    trial_BA,
                    -abs(trial_disc)
                ]
            )

        
        cur_lr, cur_zeta = from_unit(
            obs_x[-1]
        )

        

        val_acc, val_bacc, val_fm, val_probs = full_eval(
            global_model,
            X_val_tensor,
            y_val_np,
            prot_val,
            return_scores=True
        )

        val_di = (
            val_fm['DI']
            if not np.isnan(val_fm['DI'])
            else 0.0
        )

        round_history.append({
            'round': rnd,

            
            'global_acc': val_acc,
            'global_bacc': val_bacc,
            'global_di1': 1 - val_di,
            'global_spd': val_fm['SPD'],
            'global_eod': val_fm['EOD'],
            'global_aod': val_fm['AOD'],
            'global_auc': val_fm.get(
                'auc_overall',
                np.nan
            ),

            
            'mobo_lr': cur_lr,
            'mobo_zeta': cur_zeta,
        })

        print(
            f"Round {rnd}/{num_rounds} | "
            f"Val Acc={val_acc:.4f} | "
            f"Val BAcc={val_bacc:.4f} | "
            f"Val DI={val_di:.4f}"
        )

    

    g_acc, g_bacc, g_fm, g_probs = full_eval(
        global_model,
        X_test_tensor,
        y_test_np,
        prot_test,
        return_scores=True
    )

    

    final_di = (
        g_fm['DI']
        if not np.isnan(g_fm['DI'])
        else 0.0
    )

    result = {
        'selected_round': num_rounds,
        'selected_type': 'global',

        
        'final_acc': g_acc,
        'final_bacc': g_bacc,
        'final_di1': 1 - final_di,
        'final_spd': g_fm['SPD'],
        'final_eod': g_fm['EOD'],
        'final_aod': g_fm['AOD'],

        
        'final_fprd': g_fm['FPRD'],
        'final_tnrd': g_fm['TNRD'],
        'final_ppd': g_fm['PPD'],

       
        'tpr_priv': g_fm['tpr_priv'],
        'tpr_unpriv': g_fm['tpr_unpriv'],

        
        'fpr_priv': g_fm['fpr_priv'],
        'fpr_unpriv': g_fm['fpr_unpriv'],

        
        'tnr_priv': g_fm['tnr_priv'],
        'tnr_unpriv': g_fm['tnr_unpriv'],

        
        'fnr_priv': g_fm['fnr_priv'],
        'fnr_unpriv': g_fm['fnr_unpriv'],

        
        'precision_priv': g_fm['precision_priv'],
        'precision_unpriv': g_fm['precision_unpriv'],

        
        'f1_priv': g_fm['f1_priv'],
        'f1_unpriv': g_fm['f1_unpriv'],

        
        'f1_overall': g_fm['f1_overall'],
        'precision_overall': g_fm['precision_overall'],

        
        'auc_overall': g_fm.get(
            'auc_overall',
            np.nan
        ),
        'auc_priv': g_fm.get(
            'auc_priv',
            np.nan
        ),
        'auc_unpriv': g_fm.get(
            'auc_unpriv',
            np.nan
        ),

        
        'final_probs': g_probs,
        'y_test': y_test_np,
        'prot_test': prot_test,

        
        'round_history': round_history,

     
        'final_mobo_lr': cur_lr,
        'final_mobo_zeta': cur_zeta,
    }

   

    if SAVE_MODEL_WEIGHTS:
        result['state_dict'] = global_model.state_dict()

    return result


print('run_fairtrade_pipeline() ready.')
```

```python
def get_csv_path(dataset_name):
    return os.path.join(CSV_DIR, f'{dataset_name.lower()}_processed.csv')


def load_dataset(dataset_name, protected_attr):
    path = get_csv_path(dataset_name)
    df = pd.read_csv(path)
    target_col = TARGET_COL_OVERRIDES.get(dataset_name, TARGET_COL_DEFAULT)
    feature_cols = [c for c in df.columns if c not in [target_col, protected_attr]]
    numeric_cols = df[feature_cols].select_dtypes(include=['int64', 'float64']).columns
    scaler = MinMaxScaler()
    df = df.copy()
    df[numeric_cols] = scaler.fit_transform(df[numeric_cols])
    df[target_col] = df[target_col].astype(int)
    df[protected_attr] = df[protected_attr].astype(int)
    return df, target_col


def _norm(v):
    return '' if v is None else str(v)


def load_completed_keys():
    if not os.path.exists(MASTER_LOG_PATH):
        return set()
    log = pd.read_csv(MASTER_LOG_PATH)
    keys = set(zip(
        log['dataset'], log['protected_attr'], log['experiment_type'], log['method'], log['variant'],
        log['param_name'].fillna('').astype(str), log['param_value'].fillna('').astype(str),
        log['seed'].astype(str)
    ))
    return keys


def append_master_row(row: dict):
    df_row = pd.DataFrame([row])
    header = not os.path.exists(MASTER_LOG_PATH)
    df_row.to_csv(MASTER_LOG_PATH, mode='a', header=header, index=False)


def append_history_rows(dataset, protected_attr, experiment_type, method, variant,
                         param_name, param_value, seed, round_history):
    rows = []
    for rh in round_history:
        r = dict(rh)
        r.update({
            'dataset': dataset, 'protected_attr': protected_attr,
            'experiment_type': experiment_type, 'method': method, 'variant': variant,
            'param_name': _norm(param_name), 'param_value': _norm(param_value),
            'seed': seed,
        })
        rows.append(r)
    df_rows = pd.DataFrame(rows)
    header = not os.path.exists(HISTORY_LOG_PATH)
    df_rows.to_csv(HISTORY_LOG_PATH, mode='a', header=header, index=False)


def save_roc_data(method, dataset_label, protected_attr, seed, y_test, final_probs, prot_test,
                   final_mobo_lr=None, final_mobo_zeta=None, selected_round=None, selected_type=None,
                   experiment_type='main', param_name=None, param_value=None):
    tag = '' if experiment_type == 'main' else f'__{experiment_type}-{_norm(param_name)}{_norm(param_value)}'
    fname = f'{method}__{dataset_label}{tag}__seed{seed}.npz'
    np.savez_compressed(os.path.join(ROC_DIR, fname),
                         y_true=np.asarray(y_test), y_score=np.asarray(final_probs),
                         protected=np.asarray(prot_test),
                         method=method, final_mobo_lr=final_mobo_lr, final_mobo_zeta=final_mobo_zeta,
                         num_clients=NUM_CLIENTS, local_epochs=LOCAL_EPOCHS, batch_size=BATCH_SIZE,
                         selected_round=selected_round, selected_type=selected_type)

print('Logging helpers ready.')
```

```python
completed = load_completed_keys()

def already_done(key):
    return key in completed

def mark_done(key):
    completed.add(key)

def make_key(dataset, protected_attr, experiment_type, method, variant, param_name, param_value, seed):
    return (dataset, protected_attr, experiment_type, method, variant,
            _norm(param_name), _norm(param_value), str(seed))


def _extended_row_fields(res):
    return {
        'fprd': res['final_fprd'], 'tnrd': res['final_tnrd'], 'ppd': res['final_ppd'],
        'tpr_priv': res['tpr_priv'], 'tpr_unpriv': res['tpr_unpriv'],
        'fpr_priv': res['fpr_priv'], 'fpr_unpriv': res['fpr_unpriv'],
        'tnr_priv': res['tnr_priv'], 'tnr_unpriv': res['tnr_unpriv'],
        'fnr_priv': res['fnr_priv'], 'fnr_unpriv': res['fnr_unpriv'],
        'precision_priv': res['precision_priv'], 'precision_unpriv': res['precision_unpriv'],
        'f1_priv': res['f1_priv'], 'f1_unpriv': res['f1_unpriv'],
        'f1_overall': res['f1_overall'], 'precision_overall': res['precision_overall'],
        'auc_overall': res['auc_overall'], 'auc_priv': res['auc_priv'], 'auc_unpriv': res['auc_unpriv'],
        
        'final_mobo_lr': res['final_mobo_lr'], 'final_mobo_zeta': res['final_mobo_zeta'],
    }


for ds_name, prot_attr in zip(dataset_names, protected_attributes):
    ds_label = f'{ds_name}_{prot_attr}'
    print(f'\n=== {ds_label}  [{METHOD_NAME}] ===')

    try:
        df, target_col = load_dataset(ds_name, prot_attr)
    except FileNotFoundError:
        print(f'  !! CSV not found for {ds_name} at {get_csv_path(ds_name)} — skipping.')
        continue

    for seed in SEEDS_MAIN:
        key = make_key(ds_label, prot_attr, 'main', METHOD_NAME, 'full', None, None, seed)
        if already_done(key):
            continue
        res = run_fairtrade_pipeline(df, prot_attr, target_col, num_rounds=NUM_ROUNDS_MAIN, seed=seed)
        row = {
            'dataset': ds_label, 'protected_attr': prot_attr,
            'experiment_type': 'main', 'method': METHOD_NAME, 'variant': 'full',
            'param_name': '', 'param_value': '', 'seed': seed,
            'selected_round': res['selected_round'], 'selected_type': res['selected_type'],
            'acc': res['final_acc'], 'bacc': res['final_bacc'], 'di1': res['final_di1'],
            'spd': res['final_spd'], 'eod': res['final_eod'], 'aod': res['final_aod'],
        }
        row.update(_extended_row_fields(res))
        append_master_row(row)
        append_history_rows(ds_label, prot_attr, 'main', METHOD_NAME, 'full', None, None, seed, res['round_history'])
        if SAVE_ROC_DATA:
            save_roc_data(METHOD_NAME, ds_label, prot_attr, seed, res['y_test'], res['final_probs'], res['prot_test'],
                          final_mobo_lr=res['final_mobo_lr'], final_mobo_zeta=res['final_mobo_zeta'],
                          selected_round=res['selected_round'], selected_type=res['selected_type'],
                          experiment_type='main')
        mark_done(key)
        print(f'  [main] seed={seed}  Acc={res["final_acc"]:.4f}  1-DI={res["final_di1"]:.4f}  '
              f'lr*={res["final_mobo_lr"]:.4f}  zeta*={res["final_mobo_zeta"]:.4f}')

print('\nAll queued FairTrade main-experiment runs complete (or already were, and were skipped).')
print('(No sensitivity-sweep cell — see Cell 2 note: zeta is auto-tuned by MOBO, not manually swept.)')
```
